# TEXT CLASSIFICATION STEPS

1- Bag of Words
-

```python
text = [
    "I am loving the NLP class, but sometimes it feels confusing!!!",
    "NLP is a fascinating field — it deals with text, speech, and language understanding."
]
```

* Tokenization

```python
tokenized = [word_tokenize(s.lower()) for s in text]      # It splits each sentence into lowercase words (tokens).
```

* Stopword Removal

```python
stop_words = set(stopwords.words("english"))
filtered = [[w for w in words if w.isalpha() and w not in stop_words] for words in tokenized]
```

* Build Vocabulary

```python
vocabulary = sorted(set(word for sent in filtered for word in sent))    # It collects all unique words from the filtered sentences and sorts them to build the vocabulary. 

print("Vocabulary:", vocabulary)
```

* Bag of words

```python
bow_matrix = [[words.count(word) for word in vocabulary] for words in filtered]   # It creates the Bag of Words matrix by counting how many times each vocabulary word appears in every filtered sentence. 
```

```python
print("Texts:")
for s in text:
    print("-", s)

print("\nVocabulary:", vocabulary)

print("\nBag of Words Matrix:")
for vec in bow_matrix:
    print(vec)
```

2- TF-IDF
-

```python
import math
from collections import Counter
```

```python
docs = [
    "I am loving the NLP class, but sometimes it feels confusing!!!",
    "NLP is a fascinating field — it deals with text, speech, and language understanding."
]
```

- Preprocess

```python
tokenized_docs = []
for d in docs:
    text = d.lower().replace("—", "").replace(",", "").replace("!", "").replace(".", "")
    words = text.split()
    tokenized_docs.append(words)
```

- Calculate TF

```python
N = len(tokenized_docs)
tf = Counter()
for doc in tokenized_docs:
    for term in set(doc):
        tf[term] += 1
```

- Calculate IDF

```python
idf = {}
for term, count in tf.items():
    idf[term] = math.log(N / count)
```

- Calculate TF-IDF

```python
tfidf_all = []
for doc in tokenized_docs:
    tf = Counter(doc)
    doc_len = len(doc)
    tfidf = {}
    for term, count in tf.items():
        tf_val = count / doc_len
        tfidf[term] = tf_val * idf[term]
    tfidf_all.append(tfidf)
```

- Result

```python
for i, tfidf in enumerate(tfidf_all):
    print(f"Doc {i} TF-IDF:")
    for term, value in tfidf.items():
        print(f"  {term}: {value:.4f}")
    print()
```

3- Word Embedding
-

```python
from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize
import nltk
```

- Tokenization

```python
nltk.download('punkt')
```

```python
text = [
    "I am loving the NLP class, but sometimes it feels confusing!!!",
    "NLP is a fascinating field — it deals with text, speech, and language understanding."
]
```

```python
nltk.download('punkt_tab')
```

```python
tokenized_sentences = [word_tokenize(sentence.lower()) for sentence in text]
print("Tokenized Sentences:")
print(tokenized_sentences)
```

- Train Word2vec Model

```python
model = Word2Vec(
    sentences=tokenized_sentences,
    vector_size=100,   # each word represented by 100-d vector
    window=5,          # context window
    min_count=1,       # include all words
    sg=1,              # 1 = Skip-gram, 0 = CBOW
    workers=4          # number of CPU cores
)
```

```python
model.train(tokenized_sentences, total_examples=len(tokenized_sentences), epochs=10)
```

- Word Embedding

```python
print("\nVector for 'nlp':")
print(model.wv['nlp'])
```

```python
print("\nMost similar words to 'nlp':")
print(model.wv.most_similar('nlp'))
```

```python
model.save("my_word2vec.model")
loaded_model = Word2Vec.load("my_word2vec.model")
print("\nModel saved and loaded successfully!")
```
